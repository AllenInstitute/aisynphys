```python
import multiprocessing as mp


def run_one(expt):
    out_file = OUTPUT_DIR / f"exp_{expt.id:05d}.h5"
    writer = ExperimentWriter(out_file)
    writer.process_experiment(expt)


def process_all_experiments(experiments):

    for expt in experiments:
        p = mp.Process(target=run_one, args=(expt,))
        p.start()
        p.join()
```
